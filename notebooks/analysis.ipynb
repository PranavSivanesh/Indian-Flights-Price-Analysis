```python
current_airlines = ['SpiceJet', 'Vistara', 'Indigo', 'Air India', 'StarAir']
df_filtered = df_price[df_price['airline'].isin(current_airlines)]
print(df_filtered['airline'].value_counts())
print(df_filtered.shape)
```

```python
class_airline_price = df_filtered.groupby(['class', 'airline'])['price'].mean().sort_values()
class_airline_price
```

```python
df_filtered['via_city']= df_filtered['stops'].str.extract(r'Via (\w+)')
df_filtered['via_city'].value_counts()
```

```python
code_to_city = {'IXU': 'Aurangabad', 'IDR': 'Indore', 'PAT': 'Patna', 'MYQ': 'Mysore',
    'KLH': 'Kolhapur', 'JGB': 'Jagdalpur', 'JRG': 'Jharsuguda', 'STV': 'Surat',
    'BBI': 'Bhubaneswar', 'IXE': 'Mangalore', 'VTZ': 'Vishakhapatnam',
    'RPR': 'Raipur', 'NDC': 'Nanded', 'GOP': 'Gorakhpur', 'NAG': 'Nagpur',
    'IXR': 'Ranchi', 'GAY': 'Gaya', 'GAU': 'Guwahati', 'HYD': 'Hyderabad'}
df_filtered['via_city_clean'] = df_filtered['via_city'].replace(code_to_city)
df_filtered['via_city_clean'].value_counts()
```

```python
via_city_price = df_filtered[df_filtered['via_city_clean'].notna()].groupby('via_city_clean')['price'].agg(['mean', 'count']).sort_values('mean')
via_city_price
```

```python
df_filtered[df_filtered['via_city_clean'].notna()].groupby('via_city_clean')['class'].value_counts()
```

```python
via_city_price_by_class = df_filtered[df_filtered['via_city_clean'].notna()].groupby(['class', 'via_city_clean'])['price'].agg(['mean', 'count']).sort_values(['class', 'mean'])
via_city_price_by_class
```
